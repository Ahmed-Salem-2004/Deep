```python
import numpy as np

# Activation functions
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

def tanh(x):
    return np.tanh(x)
```

```python
# LSTM Parameters
# Forget gate weights
Wf_x = 0.5
Wf_h = 0.1
bf = 0

# Input gate weights
Wi_x = 0.6
Wi_h = 0.2
bi = 0

# Candidate state weights
Wc_x = 0.7
Wc_h = 0.3
bc = 0

# Output gate weights
Wo_x = 0.8
Wo_h = 0.4
bo = 0
```

```python
# Initial states
h = 0
C = 0

# Input sequence
X = [1, 2, 3]
```

```python
#LSTM Forward Pass
print("==== LSTM Forward Pass ====\n")

for t, x in enumerate(X, start=1):

    print(f"Time Step {t} , Input = {x}")

    # Forget gate
    f = sigmoid(Wf_x*x + Wf_h*h + bf)

    # Input gate
    i = sigmoid(Wi_x*x + Wi_h*h + bi)

    # Candidate cell state
    C_tilde = tanh(Wc_x*x + Wc_h*h + bc)

    # Cell state update
    C = f*C + i*C_tilde

    # Output gate
    o = sigmoid(Wo_x*x + Wo_h*h + bo)

    # Hidden state
    h = o * tanh(C)

    print("Forget gate:", round(f,3))
    print("Input gate:", round(i,3))
    print("Candidate state:", round(C_tilde,3))
    print("Cell state:", round(C,3))
    print("Output gate:", round(o,3))
    print("Hidden state:", round(h,3))
    print()

```

```python
# Predict next value (t=4)
x = 4

f = sigmoid(Wf_x*x + Wf_h*h + bf)
i = sigmoid(Wi_x*x + Wi_h*h + bi)
C_tilde = tanh(Wc_x*x + Wc_h*h + bc)
C = f*C + i*C_tilde
o = sigmoid(Wo_x*x + Wo_h*h + bo)
h = o * tanh(C)

print("==== Prediction Step ====")
print("h4 =", round(h,3))


# Linear output layer
Wy = 4
by = 0

y_pred = Wy*h + by

print("Predicted next value =", round(y_pred,3))
```
