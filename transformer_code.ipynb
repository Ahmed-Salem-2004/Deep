```python
import numpy as np

text = "What are the symptoms of diabetes?"
words = text.split()

print("Input Tokens:")
print(words)
print()
```

```python
embedding_size = 8
attention_dim = 4

np.random.seed(7)
length = len(words)
X = np.random.randn(length, embedding_size)

print("Embedding Shape:", X.shape)
print()
```

```python
Wq = np.random.randn(embedding_size, attention_dim)
Wk = np.random.randn(embedding_size, attention_dim)
Wv = np.random.randn(embedding_size, attention_dim)

Q = np.dot(X, Wq)
K = np.dot(X, Wk)
V = np.dot(X, Wv)

scores = np.dot(Q, K.T)
scaled_scores = scores / np.sqrt(attention_dim)

def softmax(matrix):
    shifted = matrix - np.max(matrix, axis=1, keepdims=True)
    exp_values = np.exp(shifted)
    return exp_values / np.sum(exp_values, axis=1, keepdims=True)

attention = softmax(scaled_scores)
context_vectors = np.dot(attention, V)
```

```python
print("Scaled Attention Scores:")
print(np.round(scaled_scores, 3))

print("\nAttention Weights:")
print(np.round(attention, 4))

print("\nContextual Representations:")
print(np.round(context_vectors, 4))
```
